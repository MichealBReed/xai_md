```python
import nupack
```

# Cy3

```python
mymodel = nupack.Model(material='dna', celsius=25)
```

```python
brightest = nupack.Domain('GCGCG'[::-1], name='bright')
middle = nupack.Domain('TCAAG'[::-1], name='middle')
dimmest = nupack.Domain('CCCCG'[::-1], name='dim')
polyT = nupack.Domain('T5', name='body')
```

## Short Sequenece (10nt)

```python
bright_strand = nupack.TargetStrand([polyT,brightest], name='brightest_strand')
bright_strand_comp = nupack.TargetStrand([~brightest,~polyT],name = 'brightest_strand_comp')

middle_strand = nupack.TargetStrand([polyT,middle], name='middle_strand')
middle_strand_comp = nupack.TargetStrand([~middle,~polyT],name = 'middle_strand_comp')

dim_strand = nupack.TargetStrand([polyT,dimmest], name = 'dimmest_strand')
dim_strand_comp = nupack.TargetStrand([~dimmest,~polyT],name = 'dimmest_strand_comp')
```

```python
bright_complex = nupack.TargetComplex([bright_strand,bright_strand_comp], '(10+)10', name = 'bright_complimentary')
middle_complex = nupack.TargetComplex([middle_strand,middle_strand_comp], '(10+)10', name = 'middle_complimentary')
dim_complex = nupack.TargetComplex([dim_strand,dim_strand_comp], '(10+)10', name = 'dim_complimentary')
```

```python
bright_tube = nupack.TargetTube(on_targets={bright_complex : 1e-8},
                               name='bright_tube',
                               off_targets=nupack.SetSpec(max_size=3))

middle_tube = nupack.TargetTube(on_targets={middle_complex : 1e-8},
                               name='middle_tube',
                               off_targets=nupack.SetSpec(max_size=3))

dim_tube = nupack.TargetTube(on_targets={dim_complex : 1e-8},
                               name='dim_tube',
                               off_targets=nupack.SetSpec(max_size=3))
```

```python
mytubes = [bright_tube, middle_tube, dim_tube]
```

```python
my_design = nupack.tube_design(tubes=mytubes,
                              model=mymodel)
```

```python
my_results = my_design.run(trials=1)
```
