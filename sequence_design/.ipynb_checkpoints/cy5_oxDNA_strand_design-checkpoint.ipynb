```python
import nupack
```

# Cy5

```python
mymodel = nupack.Model(material='dna', celsius=25)
```

```python
brightest = nupack.Domain('GCGCG'[::-1], name='bright')
dimmest = nupack.Domain('CCTTC'[::-1], name='dim')
```

```python
linker = nupack.Domain('T5', name='linker')
trinucleotide_region = nupack.Domain('N5',name='trinucleotide_stand_in')
spacer = nupack.Domain('T15',name='spacer')
```

```python
bright_strand = nupack.TargetStrand([linker,trinucleotide_region,spacer,brightest], name='brightest_strand')
dim_strand = nupack.TargetStrand([linker,trinucleotide_region,spacer,dimmest], name = 'dimmest_strand')
```

```python
bright_complex = nupack.TargetComplex([bright_strand], '.30', name = 'bright')
dim_complex = nupack.TargetComplex([dim_strand], '.30', name = 'dim')
```

```python
bright_tube = nupack.TargetTube(on_targets={bright_complex : 1e-8},
                               name='bright_tube',
                               off_targets=nupack.SetSpec(max_size=3))

dim_tube = nupack.TargetTube(on_targets={dim_complex : 1e-8},
                               name='dim_tube',
                               off_targets=nupack.SetSpec(max_size=3))
```

```python
mytubes = [bright_tube, dim_tube]
```

```python
my_design = nupack.tube_design(tubes=mytubes,
                              model=mymodel)
```

```python
my_results = my_design.run(trials=1)
```

```python
my_results[0].save_text('cy5_strand_design.txt')
```
