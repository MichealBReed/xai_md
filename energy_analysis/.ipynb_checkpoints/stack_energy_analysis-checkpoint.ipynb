```python
import numpy as np
import matplotlib.pyplot as plt
import json
```

# Cy3

## Bright

```python
cy3_bright_bexc = {}

with open('../../data/oxDNA/cy3/bright/cy3_bright_BEXC.json','r') as f:
    cy3_bright_bexc = json.load(f)


cy3_bright_crstck = {}

with open('../../data/oxDNA/cy3/bright/cy3_bright_CRSTCK.json','r') as f:
    cy3_bright_crstck = json.load(f)


cy3_bright_cxstck = {}

with open('../../data/oxDNA/cy3/bright/cy3_bright_CXSTCK.json','r') as f:
    cy3_bright_cxstck = json.load(f)


cy3_bright_dh = {}

with open('../../data/oxDNA/cy3/bright/cy3_bright_DH.json','r') as f:
    cy3_bright_dh = json.load(f)


cy3_bright_fene = {}

with open('../../data/oxDNA/cy3/bright/cy3_bright_FENE.json','r') as f:
    cy3_bright_fene = json.load(f)



cy3_bright_hb = {}

with open('../../data/oxDNA/cy3/bright/cy3_bright_HB.json','r') as f:
    cy3_bright_hb = json.load(f)



cy3_bright_nexc = {}

with open('../../data/oxDNA/cy3/bright/cy3_bright_NEXC.json','r') as f:
    cy3_bright_nexc = json.load(f)



cy3_bright_stck = {}

with open('../../data/oxDNA/cy3/bright/cy3_bright_STCK.json','r') as f:
    cy3_bright_stck = json.load(f)


cy3_bright_total = {}

with open('../../data/oxDNA/cy3/bright/cy3_bright_total.json','r') as f:
    cy3_bright_total = json.load(f)
```

```python
cy3_energies = [cy3_bright_bexc['BEXC (oxDNA su)'], cy3_bright_crstck['CRSTCK (oxDNA su)'],
                 cy3_bright_cxstck['CXSTCK (oxDNA su)'], cy3_bright_dh['DH (oxDNA su)'],
                 cy3_bright_fene['FENE (oxDNA su)'],cy3_bright_hb['HB (oxDNA su)'],
                 cy3_bright_nexc['NEXC (oxDNA su)'],cy3_bright_stck['STCK (oxDNA su)'],
                 cy3_bright_total['total (oxDNA su)']]
```

```python
plt.figure(figsize=(10,8), dpi=150)

cats = ['BEXC', 'CRSTCK', 'CXSTCK', 
       'DH', 'FENE', 'HB',
       'NEXC', 'STCK', 'total']

plt.ylabel('oxDNA Energy Units',fontsize=14)

plt.boxplot(cy3_energies, tick_labels=cats)


plt.yticks(fontsize=14)

plt.xticks(fontsize=14)


plt.savefig('cy3_energies.png')
#plt.show()
```
